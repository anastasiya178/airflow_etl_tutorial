# Data cleaning and preparation

### Managing Missing Values

For numeric data, pandas uses the floating-point value NaN (Not a number) to represent missing data. This is also called **a sentinel value **.

NAN handling methods in pandas: 

* dropna
* fillna
* isnull
* notnull

Two strategies of handling missing data: 
- filter out (remove)
- fill in


First, let's check how many values in every column
